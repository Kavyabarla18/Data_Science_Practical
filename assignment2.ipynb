```python
import pandas as pd

file_path = r"D:\Data_Science_Practical\OnlineRetail.csv"

# ISO-8859-1 (latin1) safely handles special characters like £
df = pd.read_csv(file_path, encoding='ISO-8859-1')

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
missing = df.isnull().sum().sort_values(ascending=False)

missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_percentage.round(2)
})

display(missing_summary)
```

```python

total_missing = df.isnull().sum().sum()

print("Total missing cells:", total_missing)
print("Rows containing at least one missing value:",
      df.isnull().any(axis=1).sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

duplicate_percentage = (
    duplicate_count / len(df) * 100
)

print(
    f"Duplicate percentage: {duplicate_percentage:.2f}%"
)
```

```python
duplicates = df[df.duplicated(keep=False)].sort_values(
    by=list(df.columns)
)

display(duplicates.head(20))
```

```python

print("InvoiceDate data type before conversion:")
print(df["InvoiceDate"].dtype)

print("\nQuantity data type:")
print(df["Quantity"].dtype)

print("\nUnitPrice data type:")
print(df["UnitPrice"].dtype)

print("\nCustomerID data type:")
print(df["CustomerID"].dtype)
```

```python
# Convert InvoiceDate to datetime

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print("InvoiceDate data type after conversion:")
print(df["InvoiceDate"].dtype)
```

```python
# Identify cancelled invoices

df["Is_Cancelled"] = (
    df["InvoiceNo"]
    .astype(str)
    .str.upper()
    .str.startswith("C")
)

cancellation_count = df["Is_Cancelled"].sum()

print("Cancelled transaction rows:", cancellation_count)

print(
    f"Percentage of transaction rows marked as cancelled: "
    f"{cancellation_count / len(df) * 100:.2f}%"
)
```

```python
cancellation_summary = df["Is_Cancelled"].value_counts()

display(cancellation_summary)
```

```python
negative_quantity_count = (df["Quantity"] < 0).sum()
zero_quantity_count = (df["Quantity"] == 0).sum()

print("Negative quantity rows:", negative_quantity_count)
print("Zero quantity rows:", zero_quantity_count)

print(
    f"Negative quantity percentage: "
    f"{negative_quantity_count / len(df) * 100:.2f}%"
)
```

```python
print("Negative quantity examples:")

display(
    df[df["Quantity"] < 0].head(10)
)
```

```python
negative_price_count = (df["UnitPrice"] < 0).sum()
zero_price_count = (df["UnitPrice"] == 0).sum()

print("Negative UnitPrice rows:", negative_price_count)
print("Zero UnitPrice rows:", zero_price_count)
```

```python
display(
    df[df["UnitPrice"] <= 0].head(20)
)
```

```python
retail = df.copy()

# Remove exact duplicate rows
retail = retail.drop_duplicates()

# Remove rows without essential transaction information
retail = retail.dropna(
    subset=["InvoiceNo", "StockCode", "InvoiceDate", "Quantity", "UnitPrice"]
)

# Keep positive quantity and positive price for sales analysis
sales = retail[
    (retail["Quantity"] > 0) &
    (retail["UnitPrice"] > 0) &
    (~retail["Is_Cancelled"])
].copy()

print("Original rows:", len(df))
print("Rows after duplicate removal:", len(retail))
print("Rows used for positive sales analysis:", len(sales))
```

```python
print("Cleaning summary")

cleaning_summary = pd.DataFrame({
    "Stage": [
        "Original dataset",
        "After duplicate removal",
        "Final positive-sales dataset"
    ],
    "Rows": [
        len(df),
        len(retail),
        len(sales)
    ]
})

display(cleaning_summary)
```

```python
# Calculate transaction-line revenue

sales["Revenue"] = (
    sales["Quantity"] * sales["UnitPrice"]
)

display(
    sales[
        [
            "InvoiceNo",
            "StockCode",
            "Quantity",
            "UnitPrice",
            "Revenue"
        ]
    ].head(10)
)
```

```python
print("Total positive sales revenue:")
print(f"£{sales['Revenue'].sum():,.2f}")

print("\nAverage transaction-line revenue:")
print(f"£{sales['Revenue'].mean():,.2f}")

print("\nMedian transaction-line revenue:")
print(f"£{sales['Revenue'].median():,.2f}")
```

```python
country_sales = (
    sales.groupby("Country").agg(
        Revenue=("Revenue", "sum"),
        Quantity=("Quantity", "sum"),
        Transactions=("InvoiceNo", "nunique"),
        Customers=("CustomerID", "nunique")
    ).sort_values("Revenue", ascending=False)
)

display(country_sales.head(15))
```

```python
customer_analysis = customer_sales.copy()

customer_analysis["Revenue_Segment"] = pd.qcut(
    customer_analysis["Revenue"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ],
    duplicates="drop"
)

segment_summary = (
    customer_analysis
    .groupby("Revenue_Segment", observed=False)
    .agg(
        Customers=("Revenue", "count"),
        Total_Revenue=("Revenue", "sum"),
        Average_Revenue=("Revenue", "mean"),
        Average_Orders=("Orders", "mean")
    )
)

display(segment_summary)
```

```python
analysis_date = sales["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = (
    sales.dropna(subset=["CustomerID"])
    .groupby("CustomerID")
    .agg(
        Recency=(
            "InvoiceDate",
            lambda x: (analysis_date - x.max()).days
        ),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("Revenue", "sum")
    )
)

display(rfm.head())
```

```python
print("RFM dataset shape:", rfm.shape)

display(
    rfm.describe()
)
```

```python
display(
    rfm.sort_values(
        "Monetary",
        ascending=False
    ).head(10)
)
```

```python
rfm_correlation = rfm[
    ["Recency", "Frequency", "Monetary"]
].corr()

display(rfm_correlation)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    rfm_correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("RFM Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
Q1 = sales["Revenue"].quantile(0.25)
Q3 = sales["Revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = sales[
    (sales["Revenue"] < lower_bound) |
    (sales["Revenue"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of revenue outliers:", len(outliers))
```

```python
display(
    outliers.sort_values(
        "Revenue",
        ascending=False
    )[
        [
            "InvoiceNo",
            "StockCode",
            "Description",
            "Quantity",
            "UnitPrice",
            "Revenue",
            "Country"
        ]
    ].head(20)
)
```
